**DATA ANALYSIS - EDA**

```python
f_df.head()
```

```python
#distribution of rental revenue
rental_revenue = f_df["total_rental_revenue"].sum()

print(f"Total rental revenue: ${rental_revenue:,.2f}")
```

```python
#number of rentals per customer
c_df.groupby("customer_name")["total_rentals"].sum().sort_values(ascending=False)
```

```python
#average rental duration
min_duration = f_df["average_rental_duration"].max()
max_duration = f_df["average_rental_duration"].min()
avg_duration = f_df["average_rental_duration"].mean()

print("Minimum rental duration:", min_duration)
print("Maximum rental duration:", max_duration)
print("Average rental duration:", round(avg_duration,2))
```

```python
#checking outliers
Q1 = f_df["total_rental_revenue"].quantile(0.25)
Q3 = f_df["total_rental_revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

detect_outlier = (
    (f_df["total_rental_revenue"] < lower_bound) | (f_df["total_rental_revenue"] > upper_bound)
)
print(detect_outlier.sum())
```

```python
Q1 = c_df["total_rentals"].quantile(0.25)
Q3 = c_df["total_rentals"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

detect_outlier = (
    (c_df["total_rentals"] < lower_bound) | (c_df["total_rentals"] > upper_bound)
)
print(detect_outlier.sum())
```

```python
Q1 = c_df["total_amount_spent"].quantile(0.25)
Q3 = c_df["total_amount_spent"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

detect_outlier = (
    (c_df["total_amount_spent"] < lower_bound) | (c_df["total_amount_spent"] > upper_bound)
)
print(detect_outlier.sum())
```

```python
Q1 = f_df["average_rental_duration"].quantile(0.25)
Q3 = f_df["average_rental_duration"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

detect_outlier = (
    (f_df["average_rental_duration"] < lower_bound) | (f_df["average_rental_duration"] > upper_bound)
)
print(detect_outlier.sum())
```

```python
#Trend & Pattern Analysis -rentals and revenue change over time
t_df = t_df.iloc[::-1]

fig = px.line(
    t_df,
    x="rental_month",
    y=["total_rentals", "total_revenue"],
    title="Rental & Revenue Over Time",
    labels={
        "rental_month": "Month",
        "value": "Total",
        "variable": "Metric"
    },
    markers=True
)

fig.show()
```

```python
#best and worst month
worst_month = t_df.loc[t_df["total_revenue"].idxmin(), "rental_month"]
best_month = t_df.loc[t_df["total_revenue"].idxmax(), "rental_month"]

print("Worst month:", worst_month)
print("Best month:", best_month)
```

```python
#highest rental activity over time
best_rental_month = t_df.loc[t_df["total_rentals"].idxmax(), "rental_month"]

print("Month with highest rental:", best_rental_month)
```

```python
#customer insight - average spending per customer
customer_avg = (
    c_df.groupby("customer_name").agg(
        total_amount_spent=("total_amount_spent", "sum"),
        total_rentals=("total_rentals", "sum")
    )
)

customer_avg["avg_spent_per_rental"] = (customer_avg["total_amount_spent"] / customer_avg["total_rentals"])

customer_avg.sort_values(
    "avg_spent_per_rental",
    ascending=False
).head(10)
```

```python
c_df["first_rental_date"] = pd.to_datetime(c_df["first_rental_date"])
c_df["last_rental_date"] = pd.to_datetime(c_df["last_rental_date"])
```

```python
print(c_df[["first_rental_date", "last_rental_date"]].dtypes)
```

```python
#average time gap between rentals
average_gap_days = (
    c_df["last_rental_date"] - c_df["first_rental_date"]
).dt.days.mean()

print(f"Average time between first and last rental: {average_gap_days:.0f} days")
```

```python
#film and category insight - film with highest revenue per rental 
f_df.groupby("film_title")["total_rental_revenue"].sum().sort_values(ascending=False).head(5)
```

```python
#category with the highest average
f_df.groupby("category")["average_revenue_per_rental"].sum().sort_values(ascending=False)
```

```python
#staff insight - most efficient staff
s_df.groupby("staff_name")["total_revenue"].sum().sort_values(ascending=False)
```

```python
#geographic insights - city and country with highest porfit
c_df.groupby(["city", "country"])["total_amount_spent"].sum().sort_values(ascending=False)
```

```python
#region with the fastest growth
c_df.groupby(["country"])["total_rentals"].sum().sort_values(ascending=False)
```

```python
#Do customer preferences differ by region?
c_preference = (c_df.groupby(["country", "top_category"])["total_rentals"].sum())

c_preference.sort_values(ascending=False).head(20).reset_index()
```
